```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))

import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from src.model import get_model
from src.gradcam import GradCAM, overlay_heatmap
from src.inference import RetinalDiagnosticEngine
from src.data_preprocessing import DR_CLASSES
```

```python
engine = RetinalDiagnosticEngine(model_path=None, img_size=128)
sample_path = '../data/aptos/train_images/000c1434d8d7.png'
report = engine.diagnose(sample_path, compute_faithfulness=False)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(report['raw_image_rgb'])
axes[0].set_title('Raw Fundus')
axes[0].axis('off')
axes[1].imshow(report['heatmap'], cmap='jet')
axes[1].set_title('Grad-CAM Heatmap (Conv 5)')
axes[1].axis('off')
axes[2].imshow(report['overlay_raw'])
axes[2].set_title(f'Diagnosis: {report["grade_name"]} ({report["confidence"]*100:.1f}%)')
axes[2].axis('off')
plt.tight_layout()
plt.show()
```
