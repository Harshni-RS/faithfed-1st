# 05: Faithfulness Evaluation & SSIM Explanation Drift

This notebook demonstrates:
1. Insertion and Deletion curves for Grad-CAM truthfulness verification.
2. Computing Insertion AUC, Deletion AUC, and composite Faithfulness Score.
3. Tracking SSIM explanation drift across private training.


```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))

import matplotlib.pyplot as plt
from src.inference import RetinalDiagnosticEngine
from src.faithfulness import compute_insertion_deletion_auc, compute_ssim_explanation_drift
```

```python
engine = RetinalDiagnosticEngine(model_path=None, img_size=128)
sample_path = '../data/aptos/train_images/000c1434d8d7.png'
report = engine.diagnose(sample_path, compute_faithfulness=True)
f_res = report['faithfulness']

plt.figure(figsize=(7, 4))
plt.plot(f_res['step_fractions'], f_res['insertion_curve'], 'g-o', label=f'Insertion AUC: {f_res["insertion_auc"]:.3f}')
plt.plot(f_res['step_fractions'], f_res['deletion_curve'], 'r--s', label=f'Deletion AUC: {f_res["deletion_auc"]:.3f}')
plt.xlabel('Fraction of Pixels Modified')
plt.ylabel('Confidence')
plt.title('Insertion/Deletion Faithfulness Verification')
plt.legend()
plt.grid(True, linestyle=':')
plt.show()
```
